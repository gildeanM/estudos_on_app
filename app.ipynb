```python
import panel as pn

pn.extension('tabulator', sizing_mode="stretch_width")


def tela_usuario():
    return pn.Column(
        pn.pane.Markdown("# 👥 Gestão de Usuários"),
        pn.pane.Markdown("### Aqui ficará o CRUD de Usuários (Inclusão, Edição, Deleção e Filtros)."),
        pn.pane.Alert("Tela em esqueleto - Aguardando implementação da Sprint 1 / Dia 2", alert_type="warning"),
        sizing_mode="stretch_width"
    )


def tela_grupo():
    return pn.Column(
        pn.pane.Markdown("# 🏫 Gestão de Grupos de Estudo"),
        pn.pane.Markdown("### Aqui ficará o CRUD de Grupos e controle de membros."),
        pn.pane.Alert("Tela em esqueleto - Aguardando implementação da Sprint 1 / Dia 2", alert_type="warning"),
        sizing_mode="stretch_width"
    )

def tela_encontro():
    return pn.Column(
        pn.pane.Markdown("# 📅 Gestão de Encontros"),
        pn.pane.Markdown("### Aqui ficará o CRUD de Encontros (Remotos e Presenciais)."),
        pn.pane.Alert("Tela em esqueleto - Aguardando implementação da Sprint 1 / Dia 4", alert_type="warning"),
        sizing_mode="stretch_width"
    )

def tela_mensagem():
    return pn.Column(
        pn.pane.Markdown("# 💬 Histórico de Mensagens"),
        pn.pane.Markdown("### Aqui ficará a listagem e moderação de mensagens enviadas."),
        pn.pane.Alert("Tela em esqueleto - Aguardando implementação da Sprint 1 / Dia 4", alert_type="warning"),
        sizing_mode="stretch_width"
    )    

def tela_relatorios():
    return pn.Column(
        pn.pane.Markdown("# 📊 Relatórios e Indicadores"),
        pn.pane.Markdown("### Aqui ficarão os gráficos analíticos gerados com psycopg2."),
        pn.pane.Alert("Tela em esqueleto - Aguardando implementação das queries na Sprint 2", alert_type="info"),
        sizing_mode="stretch_width"
)


def roteador_de_telas(aba_selecionada):
    if aba_selecionada == "Usuários":
        return tela_usuario()
    elif aba_selecionada == "Grupos":
        return tela_grupo()
    elif aba_selecionada == "Encontros":
        return tela_encontro()
    elif aba_selecionada == "Mensagens":
        return tela_mensagem()
    elif aba_selecionada == "Relatórios":
        return tela_relatorios()
    return pn.pane.Markdown("# ❌ Erro: Tela não encontrada.")


menu_navegacao = pn.widgets.RadioButtonGroup(
    name="Menu Principal",
    options=["Usuários", "Grupos", "Encontros", "Mensagens", "Relatórios"],
    button_type="success",
    orientation="vertical",
    sizing_mode="stretch_width"
)

conteudo_dinamico = pn.bind(roteador_de_telas, menu_navegacao)

template = pn.template.FastListTemplate(
    title="Estudos On - Painel de Gestão",
    sidebar=[
        pn.pane.Markdown("## 📌 Navegação"),
        menu_navegacao,
        pn.layout.Divider(),
        pn.pane.Markdown("💡 *Dica: Use este painel para alternar entre os módulos do sistema de forma instantânea.*")

    ],
    main=[conteudo_dinamico],
    accent_base_color="#2ecc71",
    header_background="#2c3e50"
)


template.servable()
```
