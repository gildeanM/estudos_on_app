```python
from database.connection import get_psycopg2_connection

try:
    # 1. Tenta obter a conexão do psycopg2
    conexao = get_psycopg2_connection()
    cursor = conexao.cursor()
    
    # 2. Executa uma query de teste simples que não precisa de tabelas
    cursor.execute("SELECT version();")
    versao_postgres = cursor.fetchone()
    
    print("✅ Conexão com PSYCOPG2 realizada com sucesso!")
    print(f"Versão do seu PostgreSQL: {versao_postgres[0]}")
    
    # 3. Fecha tudo de forma segura
    cursor.close()
    conexao.close()
    
except Exception as error:
    print("❌ Erro ao conectar via PSYCOPG2:")
    print(error)
```

```python
from database.connection import get_sqlalchemy_session
from sqlalchemy import text

try:
    # 1. Abre a sessão do SQLAlchemy
    sessao = get_sqlalchemy_session()
    
    # 2. Executa um comando simples usando a sessão
    resultado = sessao.execute(text("SELECT 1;")).fetchone()
    
    print("✅ Conexão com SQLALCHEMY realizada com sucesso!")
    print(f"Resultado do teste (deve ser 1): {resultado[0]}")
    
    # 3. Fecha a sessão
    sessao.close()
    
except Exception as error:
    print("❌ Erro ao conectar via SQLALCHEMY:")
    print(error)
```
